```python
import cv2
from google.colab.patches import cv2_imshow

pt_x = 169
pt_y = 118
img = cv2.imread(image_path + "Lenna.jpg")

# BGR彩色轉成灰階GRAY
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
print("GRAY Color Space")
cv2_imshow(img_gray)

px = img_gray[pt_x, pt_y]
print(f"Gray Color 通道值 = {px}")

# 灰階GRAY轉成BGR彩色
img_color = cv2.cvtColor(img_gray, cv2.COLOR_GRAY2BGR)
print("BGR Color Space")
cv2_imshow(img_color)

px = img_color[pt_x, pt_y]
print(f"BGR Color 通道值 = {px}")
```

```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

img = cv2.imread(image_path + "mutishapes1.jpg")
print("BGR Color Space")
cv2_imshow(img)

img_hsv = cv2.cvtColor(img, cv2.COLOR_BGR2HSV)

# define range of blue color in HSV
lower_blue = np.array([110, 50, 50])
upper_blue = np.array([130, 255, 255])

# Threshold the HSV image to get only blue colors
mask = cv2.inRange(img_hsv, lower_blue, upper_blue)

# Bitwise-AND mask and original image
res = cv2.bitwise_and(img, img, mask=mask)

print("HSV Color Space")
cv2_imshow(img_hsv)
print("Mask")
cv2_imshow(mask)
print("Result")
cv2_imshow(res)
```
