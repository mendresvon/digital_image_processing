```python
import cv2
from google.colab.patches import cv2_imshow
img = cv2.imread(image_path + "view.jpg")       # BGR 讀取
cv2_imshow(img)
img_gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)  # BGR 轉 GRAY
# img_rgb = cv2.cvtColor(img, 4)  # BGR 轉 RBG
cv2_imshow(img_gray)
```

```python
import cv2
from google.colab.patches import cv2_imshow
img = cv2.imread(image_path + "view.jpg")           # BGR讀取
cv2_imshow(img)
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)      # BGR轉RGB
cv2_imshow(img_rgb)
img_bgr = cv2.cvtColor(img_rgb, cv2.COLOR_RGB2BGR)  # RGB轉BGR
cv2_imshow(img_bgr)
```
